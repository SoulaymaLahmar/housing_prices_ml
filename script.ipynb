```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data = pd.read_csv('housing.csv')
```

```python
data
```

```python
data.info()
```

```python
data.dropna(inplace=True)
```

```python
data.info()
```

```python
from sklearn.model_selection import train_test_split

x = data.drop(['median_house_value'], axis=1)
y = data['median_house_value']
```

```python
x
```

```python
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# the full train dataset (x and y combined) 
train_data = x_train.join(y_train)
```

```python
train_data
```

```python
train_data.hist(figsize=(12, 10))
```

```python
train_data.corr(numeric_only=True)
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(train_data.corr(numeric_only=True), annot=True, cmap='coolwarm')
```

```python
train_data["total_rooms"] = np.log(train_data["total_rooms"] + 1 )
train_data["total_bedrooms"] = np.log(train_data["total_bedrooms"] + 1 )
train_data["population"] = np.log(train_data["population"] + 1 )
train_data["households"] = np.log(train_data["households"] + 1 )
```

```python
train_data.hist(figsize=(12, 10))
```

```python
train_data.ocean_proximity.value_counts()
```

```python
train_data= train_data.join(pd.get_dummies(train_data.ocean_proximity, dtype=int)).drop(['ocean_proximity'], axis=1)
```

```python
train_data
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(train_data.corr(numeric_only=True), annot=True, cmap='coolwarm')
```

```python
# the houses near the ocean are more expensive than the houses far away from the ocean. 
plt.figure(figsize=(12, 10))
sns.scatterplot(data=train_data, x='longitude', y='latitude', hue='median_house_value', palette='coolwarm', alpha=0.5)
```

```python
train_data['bedroom_ratio'] = train_data['total_bedrooms'] / train_data['total_rooms']
train_data['household_ratio'] = train_data['total_rooms'] / train_data['households']
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(train_data.corr(numeric_only=True), annot=True, cmap='coolwarm')
```

```python
from sklearn.linear_model import LinearRegression

# we did this again because we added new features to the train_data dataframe.
x_train, y_train = train_data.drop(['median_house_value'], axis=1), train_data['median_house_value']

reg = LinearRegression()
reg.fit(x_train, y_train)
```

```python
test_data = x_test.join(y_test)

test_data["total_rooms"] = np.log(test_data["total_rooms"] + 1 )
test_data["total_bedrooms"] = np.log(test_data["total_bedrooms"] + 1 )
test_data["population"] = np.log(test_data["population"] + 1 )
test_data["households"] = np.log(test_data["households"] + 1 )

test_data= test_data.join(pd.get_dummies(test_data.ocean_proximity, dtype=int)).drop(['ocean_proximity'], axis=1)

test_data['bedroom_ratio'] = test_data['total_bedrooms'] / test_data['total_rooms']
test_data['household_ratio'] = test_data['total_rooms'] / test_data['households']

x_test, y_test = test_data.drop(['median_house_value'], axis=1), test_data['median_house_value']
```

```python
reg.score(x_test, y_test)
```

```python
from sklearn.ensemble import RandomForestRegressor
forest = RandomForestRegressor()
forest.fit(x_train, y_train)
```

```python
forest.score(x_test, y_test)
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [100, 200, 300],
    'max_features': [8, 12, 20]
}
grid_search = GridSearchCV(forest, param_grid, cv=5, scoring='neg_mean_squared_error', return_train_score=True)
grid_search.fit(x_train, y_train)
```

```python
best_forest = grid_search.best_estimator_
```

```python
best_forest.score(x_test, y_test)
```
